```python
# Importing librarys

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
from matplotlib import colormaps
import seaborn as sns
%matplotlib inline
```

```python
# Importing the data frames that Tanja generated 14/07/2024

# Monut google drive in colab
from google.colab import drive
drive.mount('/content/drive')

# Import
df = pd.read_excel('/content/drive/MyDrive/Colab Notebooks/df_stat.xlsx')
df.head()
```

```python
# Creating a stacked bar diagram

# Group the data
df_agg = df.groupby(['Case', 'URL'])['Case'].value_counts().unstack().fillna(0)

# Plot the data
df_agg.plot(kind = 'bar', stacked = True, cmap ='tab20b')

# Make it look nice
plt.legend(bbox_to_anchor=(1.04, 0.5), loc="center left")
plt.title('Number of images per condition and source')
plt.xlabel('')
plt.xticks(rotation = 45)
plt.ylabel('Number of images')

# Save as pdf and download
plt.savefig('004_AnneAst_NumberOfCasesPerCondition.png', bbox_inches = 'tight')
```
